## About

This project focuses on predicting the selling price of used cars using machine learning. The dataset contains information about cars such as car name, year, present price, kilometers driven, fuel type, seller type, transmission, and previous owner information.

A Random Forest Regression model is used to learn the relationship between these features and the car's selling price. The model's performance is evaluated using MAE, MSE, RMSE, and R² Score.


## Objective

The main objective of this project is to build a machine learning regression model that can predict the selling price of a used car based on its available features.

The project aims to:

* Perform data exploration and preprocessing.
* Prepare categorical and numerical features for machine learning.
* Train a Random Forest Regression model.
* Predict car selling prices.
* Evaluate model performance using MAE, MSE, RMSE, and R² Score.
* Analyze the difference between actual and predicted prices.

## Features and Target

The target variable for this project is `Selling_Price`, which represents the selling price of the used car.

### Features (X)

* Year

## Categorical Feature Encoding

The dataset contains categorical variables such as `Fuel_Type`, `Seller_Type`, and `Transmission`. Since machine learning models require numerical input, these categorical features are converted into numerical format using One-Hot Encoding.

`drop_first=True` is used to remove the first category from each feature and avoid redundant dummy variables.


````python
category_cols = ["Fuel_Type","Seller_Type","Transmission"]
````

````python
x = pd.get_dummies(x,columns=category_cols,drop_first=True,dtype=int)
````

````python
x
````

````python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score,root_mean_squared_error
import numpy as np
````

## Train-Test Split

The dataset is divided into training and testing sets using an 80:20 ratio. The training data is used to train the Random Forest Regression model, while the testing data is used to evaluate its performance on unseen data.

`random_state=42` is used to ensure that the same data split is obtained each time the code is executed.


````python
x_train,x_test, y_train, y_test = train_test_split(
    x,y,test_size = 0.20, random_state = 42
)
````

## Random Forest Regression Model

A Random Forest Regression model is created to predict the selling price of used cars. The model combines multiple decision trees and averages their predictions to produce the final regression output.

* `n_estimators=100` → The model uses 100 decision trees.
* `max_depth=None` → The trees can grow until the stopping conditions are met.
* `random_state=42` → Ensures reproducible results.

````python
model = RandomForestRegressor(
    n_estimators = 100,
    max_depth = None,
    random_state = 42
)
````

## Model Training

The Random Forest Regression model is trained using the training dataset. The model learns the relationship between the input features and the car's selling price.

```python id="x4k8ps"
model.fit(X_train, y_train)
```

Here, `X_train` contains the training features and `y_train` contains the corresponding selling prices.


````python
model.fit(x_train,y_train)
````

## Model Prediction

The trained Random Forest Regression model is used to predict the selling prices for the test dataset. The model generates predicted values based on the features in `X_test`.


Here, `y_pred` contains the predicted selling prices for the test data.


````python
y_pred = model.predict(x_test)
````

````python
y_pred 
````

## Actual vs Predicted Values

A DataFrame is created to compare the actual selling prices from the test dataset with the prices predicted by the Random Forest Regression model. This comparison helps in understanding how closely the model's predictions match the actual values.

````python
result = pd.DataFrame({
    "Actual" : y_test,
    "Prediction" : y_pred
})
````

````python
result
````

## Model Evaluation

The performance of the Random Forest Regression model is evaluated using four regression metrics: Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R² Score.<br>

- MAE measures the average absolute difference between actual and predicted selling prices.
- MSE calculates the average squared difference between actual and predicted values.
- RMSE is the square root of MSE and indicates the typical magnitude of prediction errors.
- R² Score measures how well the model explains the variation in the target variable.

````python
MSE = mean_squared_error(y_test,y_pred)
MAE =  mean_absolute_error(y_test,y_pred)
RMSE = np.sqrt(MSE)
r2 = r2_score(y_test,y_pred)

print("MAE :", MAE)
print("MSE :", MSE)
print("RMSE :", RMSE)
print("R2 Score :",r2)
````

## Error Analysis

An error column is created to calculate the difference between the actual and predicted selling prices. This helps identify how much the model's prediction differs from the actual selling price for each car.

```python
result["Error"] = result["Actual"] - result["Prediction"]
```

A positive error indicates that the actual selling price is higher than the predicted price, while a negative error indicates that the predicted price is higher than the actual selling price.


````python
result["Error"] = result["Actual"] - result["Prediction"]
````

````python
result["Error"]
````

### Conclusion

The Random Forest Regression model was used to predict the car's Selling Price. The model achieved an MAE of 1.47, an RMSE of 3.50, and an R² score of 52.44%. The results indicate that the model provides a reasonable baseline for predicting car selling prices, although there is scope for further improvement through hyperparameter tuning and feature engineering.
